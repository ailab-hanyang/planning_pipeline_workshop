## 1. 접근 방식의 발전

### 이론 배경

이번 워크샵에서 다룬 A*·Frenet·RRT는 **규칙 기반(rule-based)** 으로, 동작이 명확하고 검증이 쉽지만 복잡한 상호작용·예외 상황을 일일이 규칙으로 담기 어렵습니다.  
→ 데이터로부터 정책을 **학습**하는 방향으로 확장됩니다.

```python
%matplotlib inline
import matplotlib.pyplot as plt, matplotlib.patches as mpatches
fig, ax = plt.subplots(figsize=(12, 3.2)); ax.axis('off')
def box(x, label, sub, color):
    ax.add_patch(mpatches.FancyBboxPatch((x, 0), 2.6, 1.4, boxstyle='round,pad=0.02,rounding_size=0.1',
        facecolor=color, edgecolor='white', lw=2, alpha=0.95))
    ax.text(x+1.3, 0.95, label, ha='center', color='white', fontsize=11, fontweight='bold')
    ax.text(x+1.3, 0.4, sub, ha='center', color='white', fontsize=7.5, style='italic')
box(0.0, 'Rule-based', 'Day1-2: A*, Frenet, FSM', '#7f8c8d')
box(3.3, 'Imitation\nLearning', 'expert demo -> policy', '#2980b9')
box(6.6, 'Reinforcement\nLearning', 'reward -> policy', '#27ae60')
box(9.9, 'End-to-End', 'sensor -> control (NN)', '#c0392b')
for x in [3.0, 6.3, 9.6]:
    ax.annotate('', xy=(x+0.3, 0.7), xytext=(x, 0.7), arrowprops=dict(arrowstyle='-|>', color='gray', lw=2, mutation_scale=16))
ax.set_xlim(-0.3, 13); ax.set_ylim(-0.3, 1.8)
ax.set_title('Planning Approaches: Rule-based -> Learning-based', fontsize=13, fontweight='bold')
plt.tight_layout(); plt.show()
print('규칙 기반(이번 워크샵) -> 학습 기반(IL/RL) -> End-to-End 로 발전')
```

## 2. Imitation Learning (모방 학습)

### 이론 배경

전문가(사람) 주행 데이터 `(상태 → 행동)` 를 지도학습으로 모방합니다.

- **장점**: 데이터만 있으면 학습이 단순, 자연스러운 거동
- **단점**: 분포 이탈(distribution shift) — 학습에 없던 상황에서 취약 (→ DAgger 등으로 보완)
- 예: ALVINN, ChauffeurNet, 최근의 대규모 모방 정책

## 3. Reinforcement Learning (강화 학습)

### 이론 배경

보상(reward)을 최대화하도록 시행착오로 정책을 학습합니다.

- **장점**: 명시적 정답 없이도 목표(안전·효율) 최적화 가능
- **단점**: 보상 설계 난이도, 표본 효율·안전성 문제 → 주로 **시뮬레이터(CARLA)** 에서 학습
- 이번 Step 06의 TTC 같은 지표가 보상 설계의 재료가 됩니다.

## 4. End-to-End (E2E)

### 이론 배경

센서 입력에서 제어(또는 궤적)까지를 하나의 신경망으로 직접 매핑합니다.

- **장점**: 모듈 간 정보 손실 최소화, 데이터 스케일에 따라 성능 향상
- **단점**: 해석가능성·검증 어려움, 대량 데이터 필요
- 트렌드: BEV 인지 + 토큰화된 Planning, 대규모 멀티모달, 월드모델 기반 학습

### 정리

| 접근 | 검증성 | 데이터 | 일반화 |
|------|--------|--------|--------|
| 규칙 기반 | 높음 | 불필요 | 제한적 |
| IL | 중간 | 전문가 시연 | 분포 내 |
| RL | 낮음 | 시뮬레이션 | 보상 의존 |
| E2E | 낮음 | 대규모 | 데이터 의존 |

실무에서는 규칙 기반의 **안전 보장**과 학습 기반의 **유연성**을 결합하는 하이브리드가 일반적입니다.
